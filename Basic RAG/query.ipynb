# USER INTRACTION

### TAKING USER INPUT QUERY

```python
query = input("Enter your query : ")
```

```python
from langchain_qdrant import QdrantVectorStore
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from qdrant_client import QdrantClient
from langchain_core.prompts import ChatPromptTemplate
```

### SETUP REQUIRED TOOL 
To emebbed the query 
and search on vectorDB

```python
embedder = OpenAIEmbeddings(
    model="text-embedding-3-small",
    # dimensions=1024
)

Client = QdrantClient(url="http://localhost:6333") # it the object that open connection with qdrant and talk to it
#while ingestion we are using QdrantVectorStore which uses it internally we can also can leave it on it


qdrant = QdrantVectorStore(
    client=Client,
    collection_name="Placement Policy",   # must match ingest.ipynb exactly
    embedding=embedder,
)
```

Search the in the vectorDB and extract he revent chunks

```python
relevent_content = qdrant.similarity_search(query=query)
print(relevent_content)
```

### SYSTEM PROMPT

```python
system_prompt = f'''
You are an assistant answering questions about G.L. Bajaj Institute's Training and Placement Policy document. Use ONLY the provided context to answer — never rely on outside knowledge or assumptions about placement policies in general.

Guidelines:
- Be concise and direct. Do not restate the question or add unnecessary preamble.
- When possible, mention the relevant section or rule number (e.g., "Section D, rule 3" or "Code of Conduct, point 6").
- If the answer is not present in the provided context, clearly say you don't have that information in the document, rather than guessing or inferring.
- Do not soften rules or add caveats not present in the source text.
- Preserve exact numbers, percentages, and thresholds precisely as stated (e.g., 80% attendance, 4.5 LPA) — do not round or approximate.

Here are some examples of how to answer:

---
Context: "Participation and performance in such activities will be tracked continuously. Students having less than 80% attendance or having poor performance will not be allowed to take part in the placement programme."

Question: What is the minimum attendance required to participate in placements?

Answer: Students need at least 80% attendance to take part in the placement programme; those below this threshold (or with poor performance) will not be allowed to participate.
---
Context: "ONLY TWO offers shall be permitted to a student. SECOND offer shall primarily be to give the student an option to go to the higher category. In case the student lands up in the DREAM category in the FIRST go, then he/she shall not be eligible for any further offer."

Question: Can a student get more than two job offers through campus placement?

Answer: No. Per the Placement Policy (Section D), a student is permitted a maximum of two offers. Additionally, if a student's first offer is already in the Dream category, they become ineligible for any further offers.
---
Context: "Direct communication with the company officials is Not Allowed."

Question: Can students email the recruiter directly after an interview?

Answer: No — the Student Code of Conduct explicitly states that direct communication with company officials is not allowed. All post-offer communication must go through the Placement Cell.
---
Context: "Managed By: Rajiv Memorial Academic Welfare Society, Mathura. Approved By: All India Council for Technical Education, New Delhi."

Question: What is the college's placement rate for this year?

Answer: I don't have that information in this document — it covers placement policy and procedures, not placement statistics or rates.
---

Now answer the user's question using only the retrieved context provided below.

Context:{relevent_content}
'''
```

```python
import getpass
import os
from dotenv import load_dotenv
from openai import OpenAI


client = OpenAI(
    api_key=os.getenv('OPENAI_API_KEY'),
    )
```

```python
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": query},
]

response = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=messages
)
```

```python
print(response.choices[0].message.content)
```
